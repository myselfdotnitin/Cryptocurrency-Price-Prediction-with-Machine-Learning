```python
# run pip install yfinance in terminal first
import yfinance as yf
import pandas as pd
import datetime
from datetime import date, timedelta

today = date.today()

d1 = today.strftime("%Y-%m-%d")
end_date = d1
d2 = date.today() - timedelta(days=768)
d2 = d2.strftime("%Y-%m-%d")
start_date = d2

data = yf.download('BTC-USD',
                   start = start_date,
                   end = end_date,
                   progress = False,
                   auto_adjust = False,
                   multi_level_index= False)
data["Date"] = data.index
data = data[["Date", "Open", "High", "Low", "Close", "Adj Close", "Volume"]]
data.reset_index(drop=True, inplace=True)
```

```python
correlation = data.corr()
print(correlation["Close"].sort_values(ascending=False))

```

# Ots Library is one of the best library in python for time series analysis.
So here using otto library to predict the bitcoin prices for next 30 days.

```python
pip install autots
```

```python
from autots import AutoTS
model = AutoTS(forecast_length=30, frequency='infer', ensemble='simple')
model = model.fit(data, date_col="Date", value_col="Close", id_col=None)
prediction = model.predict()
forecast = prediction.forecast
print(forecast)
```
